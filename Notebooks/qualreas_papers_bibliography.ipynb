# Annotated Bibliography

This is an annotated bibliography of 66 works on qualitative temporal and spatial reasoning with relation algebras: Allen's interval algebra and its relatives, point and interval calculi, the Region Connection Calculus, and the constraint-based methods used to reason with them.

The `qualreas` package is based on A. J. Reich's 1994 paper, *Intervals, Points, and Branching Time* (entry **[C3]**), which extends Allen's algebra with points and branching time.

The works are grouped into nine subtopics (Section 1) and, within each group, sorted by publication date from earliest to latest (Section 3). Section 4 lists all of them in a single chronological table. Dates are the years of formal publication; where a work first circulated as a draft or technical report, or exists in several versions, the entry says so and cites the published one.
