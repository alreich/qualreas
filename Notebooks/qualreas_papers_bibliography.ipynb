# Annotated Bibliography: Papers in `Papers/AllPapers/`

Bibliography of the 67 PDFs in the `qualreas` repository's `Papers/AllPapers/` directory (63 in the main folder plus 4 in `Temporal Reasoning and NLP/`), on qualitative temporal and spatial reasoning with relation algebras (Allen's interval algebra, RCC, constraint networks, and related work).

**How it is organized.** Papers are grouped into nine subtopics (Section 1). Within each group they are sorted by publication date, earliest to latest; a single chronological list of all 67 is in Section 5. Dates are the *actual* publication years, which differ from the repo filenames in a number of cases (Section 2).

Each entry lists the repo filename so you can map it back before deleting the PDFs.

## 3. About the links

- 🟢 **free full text** (PDF or open-access page)
- 🔒 **publisher or DOI page** (may require a subscription)
- 📄 **catalog or record page** (the paper may not be downloadable there)
- **†** means I could not open that URL from my environment (blocked or unreadable). It was taken from a search result or an index page that lists it, so check it before relying on it.

DOIs come from publisher or repository pages. I could not resolve doi.org itself from my environment, so DOIs were not clicked through. For the Elsevier *Artificial Intelligence* papers I only found the DOI, so check whether ScienceDirect lets you read them for free. Where several versions exist (technical report, preprint, journal), the free one is labeled as such.

Some papers have no online copy at all (marked *No online copy found*). Several of those are in publisher or workshop archives that only a browser search will turn up.
